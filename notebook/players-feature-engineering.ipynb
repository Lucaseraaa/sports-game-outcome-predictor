# Aggiunta della rosa
Utilizzando il dataset trovato su Kaggle, aggiungo anche il valore della rosa della specifica partita disputata. Per farlo importo entrambi i csv:

```python
import pandas as pd

result_path = "static/da-result/result.csv"
rosa_path = "static/da-result/rosa.csv"
df1 = pd.read_csv(result_path)
df2 = pd.read_csv(rosa_path)
```

successivamente associo le singole partite del primo dataset al secondo, ottenendo il risultato di mio interesse.

```python
df1['Date'] = pd.to_datetime(df1['Date'])
df2['data_partita'] = pd.to_datetime(df2['data_partita'])
df_merged = pd.merge(df1, df2, left_on='Date', right_on='data_partita', how='left')
maschera = [
    str(home).lower() in str(casa).lower() 
    for home, casa in zip(df_merged['HomeTeam'], df_merged['squadra_casa'])
]
df_finale = df_merged[maschera].copy()
colonne_da_rimuovere = ['squadra_casa', 'squadra_ospite', 'data_partita']
df_finale = df_finale.drop(columns=colonne_da_rimuovere)
df_finale = df_finale.reset_index(drop=True)
df_finale.rename(
    columns={
        'valore_titolari_casa': 'HomeValue',
        'valore_titolari_ospite': 'AwayValue'
    }, 
    inplace=True
)

df_finale.head()
```

# Esportazione del dataset

```python
df_finale.to_csv("static/da-result/complete-result.csv", encoding='utf-8', index=False, header=True)
```
